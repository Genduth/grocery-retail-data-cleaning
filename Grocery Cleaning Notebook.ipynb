</div>

```python
### Find all rows where qty * unit_price does not equal line_amount.

expected_line_amount = sales_df["qty"] * sales_df["unit_price"]

calculation_errors = sales_df[
    ~np.isclose(
        sales_df["line_amount"],
        expected_line_amount,
        rtol=1e-9,
        atol=0.01
    )
][sales_columns].copy()

calculation_errors["expected_line_amount"] = (
    calculation_errors["qty"]
    * calculation_errors["unit_price"]
)

calculation_errors["difference"] = (
    calculation_errors["line_amount"]
    - calculation_errors["expected_line_amount"]
)

calculation_errors
```

```python
### Test the two clear scaling patterns before correcting any values.

quantity_scale_error = (
    sales_df["qty"].isin([1000, 2000, 3000, 4000])
    & np.isclose(
        (sales_df["qty"] / 1000) * sales_df["unit_price"],
        sales_df["line_amount"],
        rtol=1e-9,
        atol=0.01
    )
)

line_amount_scale_error = np.isclose(
    sales_df["line_amount"],
    expected_line_amount * 100,
    rtol=1e-9,
    atol=0.01
)

pd.Series({
    "all_calculation_errors": len(calculation_errors),
    "qty_values_multiplied_by_1000": int(quantity_scale_error.sum()),
    "line_amount_values_multiplied_by_100": int(line_amount_scale_error.sum())
})
```

<div style="background-color:#f3e8ff; padding:12px; border-left:5px solid #8b5cf6; border-radius:6px;">
<h3 style="margin:0;">Calculation Error Investigation</h3>
<p style="margin-bottom:0;">
All 43 calculation errors follow one of two consistent patterns. 29 quantities are 1,000 times too large while the recorded line amount matches qty / 1000 * unit_price. The other 14 rows have sensible quantities and prices but line amounts that are exactly 100 times too large. This provides enough evidence to correct the affected field in each group.
</p>
</div>

```python
### Correct only the rows that match the confirmed scaling patterns.

sales_df["qty_was_corrected"] = False
sales_df["line_amount_was_corrected"] = False

qty_corrections_made = int(quantity_scale_error.sum())
sales_df.loc[quantity_scale_error, "qty"] = (
    sales_df.loc[quantity_scale_error, "qty"] / 1000
)
sales_df.loc[quantity_scale_error, "qty_was_corrected"] = True

expected_after_qty_fix = sales_df["qty"] * sales_df["unit_price"]
line_amount_corrections_made = int(line_amount_scale_error.sum())
sales_df.loc[line_amount_scale_error, "line_amount"] = (
    expected_after_qty_fix[line_amount_scale_error]
)
sales_df.loc[line_amount_scale_error, "line_amount_was_corrected"] = True
```

```python
### Confirm every sales calculation now matches within one penny.

calculation_errors_remaining = int((
    ~np.isclose(
        sales_df["line_amount"],
        sales_df["qty"] * sales_df["unit_price"],
        rtol=1e-9,
        atol=0.01
    )
).sum())

pd.Series({
    "qty_corrections_made": qty_corrections_made,
    "line_amount_corrections_made": line_amount_corrections_made,
    "calculation_errors_remaining": calculation_errors_remaining
})
```

29 quantity values were divided by 1,000 and 14 line amount values were divided by 100 using the confirmed calculation patterns. Every remaining row now satisfies line_amount = qty * unit_price within one penny.

```python
### Flag IDs that do not exist in the Stores and Products tables.

sales_df["store_id_was_invalid"] = (
    ~sales_df["store_id"].isin(stores_df["store_id"])
)
sales_df["product_id_was_invalid"] = (
    ~sales_df["product_id"].isin(products_df["product_id"])
)

invalid_store_rows_original = int(sales_df["store_id_was_invalid"].sum())
invalid_product_rows_original = int(sales_df["product_id_was_invalid"].sum())

pd.Series({
    "invalid_store_rows": invalid_store_rows_original,
    "invalid_product_rows": invalid_product_rows_original
})
```

```python
### Review every invalid ID before deciding whether it can be recovered.

invalid_store_rows = sales_df[sales_df["store_id_was_invalid"]]
invalid_product_rows = sales_df[sales_df["product_id_was_invalid"]]

display(invalid_store_rows)
display(invalid_product_rows)
```

```python
### Recover the two invalid IDs supported by a matching duplicate transaction.

sales_df["store_id_was_repaired"] = False
sales_df["product_id_was_repaired"] = False

valid_store_for_407 = sales_df.loc[
    (sales_df["txn_id"] == 407)
    & sales_df["store_id"].isin(stores_df["store_id"]),
    "store_id"
].iloc[0]

store_repair_mask = (
    (sales_df["txn_id"] == 407)
    & ~sales_df["store_id"].isin(stores_df["store_id"])
)
sales_df.loc[store_repair_mask, "store_id"] = valid_store_for_407
sales_df.loc[sales_df["txn_id"] == 407, "store_id_was_repaired"] = True

valid_product_for_1293 = sales_df.loc[
    (sales_df["txn_id"] == 1293)
    & sales_df["product_id"].isin(products_df["product_id"]),
    "product_id"
].iloc[0]

product_repair_mask = (
    (sales_df["txn_id"] == 1293)
    & ~sales_df["product_id"].isin(products_df["product_id"])
)
sales_df.loc[product_repair_mask, "product_id"] = valid_product_for_1293
sales_df.loc[sales_df["txn_id"] == 1293, "product_id_was_repaired"] = True

store_ids_repaired = int(store_repair_mask.sum())
product_ids_repaired = int(product_repair_mask.sum())

pd.Series({
    "store_ids_repaired": store_ids_repaired,
    "product_ids_repaired": product_ids_repaired
})
```

```python
### Keep all other invalid IDs unchanged because there is no reliable replacement.

unresolved_store_rows = sales_df[
    ~sales_df["store_id"].isin(stores_df["store_id"])
]
unresolved_product_rows = sales_df[
    ~sales_df["product_id"].isin(products_df["product_id"])
]

print("Unresolved store IDs:", len(unresolved_store_rows))
display(unresolved_store_rows)

print("Unresolved product IDs:", len(unresolved_product_rows))
display(unresolved_product_rows)
```

<p style="margin-bottom:0;">
One invalid store ID and one invalid product ID were recovered from matching records with the same transaction ID. The remaining 13 store IDs and 28 product IDs use 9999, but no other field provides a reliable replacement. They have been retained and flagged rather than guessed or deleted.

```python
### Preserve repair evidence, then remove the three records that are now confirmed duplicates.

repair_flags = [
    "qty_was_corrected",
    "line_amount_was_corrected",
    "store_id_was_repaired",
    "product_id_was_repaired"
]

for flag in repair_flags:
    sales_df[flag] = (
        sales_df.groupby("txn_id")[flag]
        .transform("max")
        .astype(bool)
    )

repaired_duplicates_removed = int(
    sales_df.duplicated(subset=sales_columns).sum()
)

sales_df = sales_df.drop_duplicates(
    subset=sales_columns
).copy()

pd.Series({
    "additional_exact_duplicates_removed": repaired_duplicates_removed,
    "duplicate_transaction_ids_remaining": sales_df["txn_id"].duplicated().sum(),
    "rows_remaining": len(sales_df)
})
```

<h3 style="margin:0;">Missing Transaction Dates</h3>
</div>

```python
### Compare each missing date with the previous and next transaction dates.

sales_df["txn_date_was_missing"] = sales_df["txn_date"].isnull()

sales_sorted = sales_df.sort_values("txn_id").copy()
sales_sorted["previous_date"] = sales_sorted["txn_date"].shift(1)
sales_sorted["next_date"] = sales_sorted["txn_date"].shift(-1)

missing_date_review = sales_sorted[
    sales_sorted["txn_date"].isnull()
][
    ["txn_id", "previous_date", "txn_date", "next_date"]
].copy()

missing_date_review["safe_to_repair"] = (
    missing_date_review["previous_date"].notnull()
    & missing_date_review["previous_date"].eq(
        missing_date_review["next_date"]
    )
)

missing_date_review
```

```python
### Fill only dates where the previous and next transaction dates agree.

safe_dates = missing_date_review[
    missing_date_review["safe_to_repair"]
].set_index("txn_id")["previous_date"]

date_repair_mask = (
    sales_df["txn_date"].isnull()
    & sales_df["txn_id"].isin(safe_dates.index)
)

txn_dates_repaired = int(date_repair_mask.sum())

sales_df.loc[date_repair_mask, "txn_date"] = (
    sales_df.loc[date_repair_mask, "txn_id"]
    .map(safe_dates)
)

pd.Series({
    "transaction_dates_repaired": txn_dates_repaired,
    "missing_dates_remaining": sales_df["txn_date"].isnull().sum()
})
```

All 29 missing dates occurred between transactions with the same date. Only those matching neighbour dates were used, so no date was filled using a blind forward or backward fill.
